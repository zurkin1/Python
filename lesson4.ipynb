<div style="text-align: right;" dir="rtl">
<h2>שילוב תנאים בעזרת and</h2>
<p>עד עכשיו בדקנו תנאי אחד בכל פעם. לפעמים נרצה שהתוכנית תבצע פעולה רק אם <b>כמה</b> תנאים מתקיימים בו-זמנית.</p>
<p>לשם כך נשתמש במילה <b>and</b>. הביטוי כולו יהיה True רק אם <b>שני</b> הצדדים שלו הם True.</p>
</div>

```python
age = 15
has_ticket = True

if age >= 12 and has_ticket:
    print("Allowed")
else:
    print("Not allowed")
```

<div style="text-align: right;" dir="rtl">
<p>נסו לשנות את הערך של has_ticket ל- False ולהריץ שוב. שימו לב שגם אם הגיל מתאים, מספיק שתנאי אחד הוא False כדי שכל הביטוי יהיה False.</p>
</div>

<div style="text-align: right;" dir="rtl">
<h2>תנאי מקונן (תנאי בתוך תנאי)</h2>
<p>לפעמים נרצה לבדוק תנאי מסוים רק אחרי שתנאי אחר כבר התקיים. במקרה כזה נכתוב תנאי בתוך תנאי - מה שנקרא <b>תנאי מקונן</b>.</p>
<p>שימו לב להזחה (הרווח בתחילת השורה) - היא זו שמסמנת לפייתון שהקוד נמצא <b>בתוך</b> התנאי החיצוני:</p>
</div>

```python
is_weekend = True
is_sunny = True

if is_weekend:
    if is_sunny:
        print("Let's go to the beach!")
    else:
        print("Let's watch a movie")
else:
    print("Time to study")
```

<div style="text-align: right;" dir="rtl">
<h2>בואו נבנה את הפתרון שלב אחרי שלב</h2>
<p>קודם כל, נבדוק את בדיקת הציונים עבור מקצוע אחד בלבד (מתמטיקה), באמצעות and לבדיקת הטווח:</p>
</div>

```python
grade = 90

if grade > 95:
    print("Excellent")
elif grade >= 85 and grade <= 95:
    print("Very good!")
else:
    print("Do better next time")
```

<div style="text-align: right;" dir="rtl">
<p>אבל בתרגיל שלנו יש שני מקצועות עם סף ציונים שונה. לכן, קודם צריך לבדוק <b>מהו המקצוע</b>, ורק אז - <b>בתוך</b> אותו תנאי - לבדוק את הציון. זהו בדיוק תנאי מקונן!</p>
</div>

```python
class_exam = "Math"
grade = 90

if class_exam == "Math":
    if grade > 95:
        print("Excellent")
    elif grade >= 85 and grade <= 95:
        print("Very good!")
    else:
        print("Do better next time")
elif class_exam == "English":
    if grade > 90:
        print("Excellent")
    elif grade >= 80 and grade <= 90:
        print("Very good!")
    else:
        print("Do better next time")
else:
    print("There is no such class")
```

<div style="text-align: right;" dir="rtl">
<p>נסו לשנות את class_exam ל-"English" ואת grade לערכים שונים, והריצו מחדש. נסו גם ערך שגוי כמו class_exam = "History" ובידקו מה קורה.</p>
</div>
