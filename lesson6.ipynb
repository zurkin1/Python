<div style="text-align: right;" dir="rtl">
<h2>לולאות</h2>
<p>עד עכשיו כתבנו קוד שמבצע כל פעולה פעם אחת בלבד. לפעמים נרצה לחזור על אותה פעולה כמה פעמים - לשם כך משתמשים ב<b>לולאה</b> (loop).</p>
<p>מבנה של לולאה בפייתון:</p>
<p>for item in range(count):<br>&nbsp;&nbsp;&nbsp;&nbsp;print(item)</p>
<ul>
<li><b>count</b> - מספר הפעמים שיש לבצע את הלולאה</li>
<li><b>item</b> - משתנה שמייצג את מונה הלולאה (אפשר לבחור כל שם אחר). הוא מתחיל מ-0 ועד count-1</li>
</ul>
</div>

<div style="text-align: right;" dir="rtl">
<p>דוגמה פרטית - הלולאה תפעיל את הוראת ההדפסה 10 פעמים, והאינדקס i יפעל מ-0 עד 9 כולל.</p>
<p>שימו לב: כמו בתנאים, גם כאן ה<b>הזחה</b> (TAB / רווחים) היא זו שמסמנת לפייתון אילו שורות שייכות ללולאה.</p>
</div>

```python
for i in range(10):
    print(i)
```

<div style="text-align: right;" dir="rtl">
<h2>דוגמה - הדפסת מספרים זוגיים</h2>
<p>אפשר להשתמש במונה הלולאה בתוך חישוב. כאן, בכל סיבוב, מודפס i כפול 2:</p>
</div>

```python
for i in range(100):
    print(i * 2)
```

<div style="text-align: right;" dir="rtl">
<h2>דוגמה - שארית חלוקה ב-2</h2>
<p>בקוד הבא, בכל סיבוב מודפסת השארית של i בחלוקה ל-2 (0 או 1 בלבד):</p>
<p>אפשר גם להפעיל לולאה בתוך לולאה - נקרא לכך <b>לולאות מקוננות</b>, ונלמד עליהן בהמשך.</p>
</div>

```python
for i in range(100):
    print(i % 2)
```

<div style="text-align: right;" dir="rtl">
<h2>תרגיל - מדבקות לפי שם</h2>
<p>לקראת סיור בחברת הי-טק, יש להדפיס תג שמי. כתבו תוכנית הקולטת את שם המבקר ותאריך הביקור, ומדפיסה מדבקה - עבור 10 מבקרים:</p>
</div>

```python
for i in range(10):
    name = input("Enter full name: ")
    date = input("Enter the date dd/mm/yy: ")
    print("====================")
    print("--------------------")
    print("| ", name, " ")
    print("| ", date, " ")
    print("--------------------")
    print("====================")
```

<div style="text-align: right;" dir="rtl">
<h2>סיכום ביניים</h2>
<p>מבנה הלולאה עם אינדקס, והדפסה פשוטה (עם end כדי להדפיס הכל באותה שורה):</p>
</div>

```python
for i in range(10):
    print(i, end=" ")
```

<div style="text-align: right;" dir="rtl">
<h2>לולאות מיוחדות - range(start, stop)</h2>
<p>אפשר להשתמש בלולאת for בדרך נוספת - הגדרת ערך התחלתי:</p>
<p>for item in range(start, stop):</p>
<ul>
<li><b>start</b> - ערך התחלתי למשתנה הלולאה (כולל)</li>
<li><b>stop</b> - ערך סופי למשתנה הלולאה (<b>לא</b> כולל)</li>
<li><b>item</b> - משתנה המייצג את מונה הלולאה. אפשר לבחור בכל שם שנרצה</li>
</ul>
</div>

<div style="text-align: right;" dir="rtl">
<p>מה יודפס למסך בהינתן הקוד הבא? נחשו לפני שאתם מריצים:</p>
<p>for i in range(100, 110):<br>&nbsp;&nbsp;&nbsp;&nbsp;print(i, end=" ")</p>
</div>

```python
for i in range(100, 110):
    print(i, end=" ")
```

<div style="text-align: right;" dir="rtl">
<h2>לולאות מיוחדות - range(start, stop, step)</h2>
<p>אפשר להשתמש בלולאת for בדרך נוספת - עם קפיצה (step):</p>
<p>for item in range(start, stop, step):</p>
<ul>
<li><b>start</b> - ערך התחלתי למשתנה הלולאה (כולל)</li>
<li><b>stop</b> - ערך סופי למשתנה הלולאה (לא כולל)</li>
<li><b>step</b> - בכמה להגדיל את משתנה הלולאה בכל סיבוב. ברוב המקרים נשתמש בערך 1. המשתנה חייב להיות מספר שלם - חיובי או שלילי</li>
</ul>
</div>

<div style="text-align: right;" dir="rtl">
<p>מה יודפס למסך בהינתן הקוד הבא? נחשו לפני שאתם מריצים:</p>
<p>for i in range(100, 120, 5):<br>&nbsp;&nbsp;&nbsp;&nbsp;print(i, end=" ")</p>
</div>

```python
for i in range(100, 120, 5):
    print(i, end=" ")
```

<div style="text-align: right;" dir="rtl">
<p>מה קורה אם ה- step שלילי? מה יודפס למסך בהינתן הקוד הבא? נחשו לפני שאתם מריצים:</p>
<p>for i in range(150, 120, -5):<br>&nbsp;&nbsp;&nbsp;&nbsp;print(i, end=" ")</p>
</div>

```python
for i in range(150, 120, -5):
    print(i, end=" ")
```

<div style="text-align: right;" dir="rtl">
<h2>חידה</h2>
<p>הריצו את שני קטעי הקוד הבאים - האם התוצאה שלהם זהה או שונה?</p>
</div>

```python
for item in range(17):
    if item % 3 == 0:
        print(item)
        print("Done")
```

```python
for item in range(0, 17, 3):
    print(item)
    print("Done")
```

<div style="text-align: right;" dir="rtl">
<p>שתי הלולאות מדפיסות בדיוק את אותה תוצאה - <b>זהה</b>! זו דוגמה לכך שלעיתים אפשר להגיע לאותה תוצאה במספר דרכים שונות.</p>
</div>

<div style="text-align: right;" dir="rtl">
<h2>המשך הדגמה - שימוש במשתנים</h2>
<p>כתבו תוכנית הקולטת תו כלשהו, ומדפיסה אותו count פעמים באותה שורה:</p>
</div>

```python
count = int(input("Number of items: "))
ch = input("Enter a character: ")

for i in range(count):
    print(ch, end=" ")
```

<div style="text-align: right;" dir="rtl">
<h2>המשך הדגמה - סכום ספרות</h2>
<p>כתבו תוכנית הקולטת מספר דו-ספרתי, ומדפיסה את סכום ספרותיו.</p>
<p>איך מחלצים את הספרות של מספר דו-ספרתי?</p>
<ul>
<li>בדיקה שהמספר אכן דו-ספרתי: num &gt; 9 and num &lt; 100</li>
<li>חילוץ ספרת האחדות: num % 10 (השארית בחלוקה ל-10)</li>
<li>חילוץ ספרת העשרות: num // 10 (חלוקה שלמה ב-10)</li>
</ul>
<p>נתחיל עם מספר אחד בודד:</p>
</div>

```python
num = 34

if num > 9 and num < 100:
    n1 = num % 10    # ספרת האחדות
    n2 = num // 10   # ספרת העשרות
    print(n1 + n2)
```

<div style="text-align: right;" dir="rtl">
<p>עכשיו נשלב את הקוד בתוך לולאה, כדי לחזור על התהליך עבור count מספרים שהמשתמש מזין:</p>
</div>

```python
count = int(input("Number of items: "))

for i in range(count):
    num = int(input("Enter a number: "))
    if num > 9 and num < 100:
        n1 = num % 10
        n2 = num // 10
        print(n1 + n2)
```

<p dir="rtl">for item in range(count) - חוזר count פעמים, item מתחיל מ-0.</p>

<p dir="rtl">for item in range(start, stop, step) - כמו קודם, אך מתקדם בקפיצות של step בכל סיבוב. step יכול להיות גם שלילי.</p>
