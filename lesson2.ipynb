<div style="text-align: right;" dir="rtl">
<h2>מה זה משתנה?</h2>
<p>דמיינו שעליכם להגיע לביתו של חברכם דני.</p>
<p>בשכונה שבה הוא גר יש עוד 3 אנשים בשם דני. אתם שואלים עוברי אורח "איפה דני גר?", ובכל פעם מקבלים תשובה אחרת...</p>
<p>איך נדע בוודאות איפה דני גר?</p>
<p>התשובה: אנחנו זקוקים למזהה חד-משמעי ויחיד – <b>כתובת</b>.</p>
<p>גם במחשב, כדי לדעת איפה שמור כל ערך, צריך "כתובת". לכתובת הזו אנחנו קוראים <b>משתנה</b>.</p>
</div>

<div style="text-align: right;" dir="rtl">
<h2>הגדרה</h2>
<p><b>משתנה</b> הוא כתובת בזיכרון המחשב, שבה שמור ערך אחד.</p>
<p>לדוגמה, נשמור את הערך 365 במשתנה בשם Danny:</p>
</div>

```python
# יצירת משתנה בשם Danny ושמירת הערך 365 בתוכו
Danny = 365
print(Danny)
```

<div style="text-align: right;" dir="rtl">
<h2>טיפוסים של משתנים</h2>
<p>ערך שנשמור במשתנה יכול להיות מכמה טיפוסים שונים:</p>
<ul>
<li>מספר שלם – int (לדוגמה: 365)</li>
<li>מספר עשרוני – float (לדוגמה: 3.65)</li>
<li>מחרוזת (טקסט) – str (לדוגמה: Bob)</li>
<li>בוליאני (אמת/שקר) – bool (לדוגמה: True)</li>
</ul>
</div>

```python
# נגדיר משתנה מכל טיפוס, ונבדוק את הטיפוס שלו באמצעות type()
count = 365
pi = 3.65
name = "Bob"
is_ready = True

print(count, type(count))
print(pi, type(pi))
print(name, type(name))
print(is_ready, type(is_ready))
```

<div style="text-align: right;" dir="rtl">
<h2>המרת טיפוסים</h2>
<p>אפשר להמיר ערך מטיפוס אחד לטיפוס אחר, לפי הצורך. הריצו את הקוד הבא ובדקו האם ניחשתם נכון:</p>
</div>

```python
print(int("67"))    # ממיר טקסט למספר שלם
print(int(64.3))    # ממיר מספר עשרוני למספר שלם (החלק העשרוני נחתך)
print(float(64))    # ממיר מספר שלם למספר עשרוני
```

<div style="text-align: right;" dir="rtl">
<h2>למה בעצם צריך משתנים?</h2>
<ul>
<li>נניח ששאלנו את המשתמש מה שמו, ואנחנו רוצים להשתמש בשם הזה שוב ושוב בהודעות שונות</li>
<li>נניח שאנחנו כותבים משחק מחשב ורוצים לשמור את הניקוד של השחקן</li>
</ul>
<p>את כל אלו נוכל לשמור בתוך משתנה!</p>
</div>

```python
# נבקש מהמשתמש את שמו, נשמור אותו במשתנה, ונשתמש בו פעמיים
name = input("What is your name? ")
print("Hello", name)
print("Nice to meet you,", name)
```

<div style="text-align: right;" dir="rtl">
<h2>פעולות על משתנים</h2>
<p>אפשר לבצע חישובים על משתנים, ולעדכן את הערך השמור בהם:</p>
</div>

```python
days = 365
days = days - 1
print(days)

days = days + days - 300 + 1
print(days)
```

<div style="text-align: right;" dir="rtl">
<h2>שאלה מסכמת – החלפה בין ערכים</h2>
<p>קיימים שני משתנים ובהם ערכים:</p>
<p>num1 = 3<br>num2 = 5</p>
<p>עליכם למצוא דרך להחליף בין הערכים, כך ש-num1 יכיל את 5, ו-num2 יכיל את 3.</p>
<p><b>הדרכה:</b> השתמשו במשתנה שלישי בשם Temp.</p>
</div>

```python
num1 = 3
num2 = 5

# כתבו כאן קוד שמחליף בין הערכים של num1 ו-num2
# בעזרת משתנה שלישי בשם Temp


print("num1 =", num1)
print("num2 =", num2)
```

<div style="text-align: right;" dir="rtl">
<h2>מונה (Counter)</h2>
<p>מונה הוא משתנה המשמש לספירה – קידום (הוספה) או הורדה (החסרה) של ערכו, בדרך כלל ב-1 בכל פעם.</p>
</div>

```python
num = 0         # אתחול המונה
print(num)

num = num + 1   # קידום
print(num)

num = num - 1   # הורדה
print(num)
```

<div style="text-align: right;" dir="rtl">
<p>דוגמה למונה מהעולם האמיתי אפשר לראות באתר <a href="https://www.worldometers.info/iw/" target="_blank">מונה עולמי</a> – שם סופרים בזמן אמת אוכלוסייה, לידות ועוד.</p>
</div>

<div style="text-align: right;" dir="rtl">
<h2>צובר (Accumulator)</h2>
<p>צובר הוא משתנה שאוגר (מצטבר) בתוכו הוספה או הפחתה של ערכים שונים, לא בהכרח קבועים.</p>
</div>

```python
num = 0
num = num + 100
print(num)

num = num + 200
print(num)
```

<div style="text-align: right;" dir="rtl">
<h2>כתיבה נכונה של שמות משתנים</h2>
<ul>
<li>יש להשתמש בשם משמעותי (למשל: name, age)</li>
<li>כותבים באותיות קטנות (Lower Case)</li>
<li>אם שם המשתנה מורכב ממספר מילים, אין לשים רווחים ביניהן. לדוגמה: myName או my_name</li>
</ul>
</div>

<div style="text-align: right;" dir="rtl">
<h2>משתנים בוליאניים</h2>
<p>הריצו את הקוד הבא וצפו בפלט המתקבל:</p>
</div>

```python
print(10 > 9)
print(10 == 9)
print(10 < 9)
```

<div style="text-align: right;" dir="rtl">
<p>הערכים המודפסים הם ערכים בוליאניים – שערכם <b>True</b> (אמת) או <b>False</b> (שקר).</p>
<p>כלומר, הביטוי שבתוך ה-print נבדק, והתוצאה היא True או False בלבד.</p>
</div>

<div style="text-align: right;" dir="rtl">
<h2>טעימה לקראת השיעור הבא – תנאים</h2>
<p>תיאור מילולי של תנאי:</p>
<p>"אם אמת: בצע"</p>
</div>

```python
if True:
    print()
```

<div style="text-align: right;" dir="rtl">
<p>דוגמה נוספת:</p>
<p>"אם השמש זורחת כל בוקר: הדפס('השמש זורחת')"</p>
</div>

```python
if True:
    print("השמש זורחת")
```

<div style="text-align: right;" dir="rtl">
<p>כאשר מציבים ערך בוליאני במשתנה, אפשר להציב רק <b>True</b> או <b>False</b>.</p>
<p>מה יודפס בהרצת הקוד הבא? נסו לנחש לפני שאתם מריצים:</p>
</div>

```python
go = True
print(go)

if True:
    print("Hi-1")

if go:
    print("Hi-2")

if (9 > 10):
    print("Hi-3")
```
