<div style="text-align: right;" dir="rtl">
<h2>תנאים</h2>
<p>לפעמים נרצה שהתוכנית שלנו תתנהג אחרת בהתאם למצב מסוים. לשם כך אנחנו משתמשים ב<b>תנאים</b>.</p>
<p>לדוגמה, נרצה לבדוק כמה זיכרון (memory) יש לטלפון, ואם יש לו הרבה – להדפיס שהוא מצוין:</p>
</div>

```python
memory = 4

if memory > 2:
    print("Excellent phone")
```

<div style="text-align: right;" dir="rtl">
<p>ראינו מה קורה אם התנאי הוא <b>אמת (True)</b>.</p>
<p>אבל מה קורה במקרה שהתנאי הוא <b>שקר (False)</b>? מה עושים?</p>
<p>לשם כך נשתמש במילה <b>else</b> – שמריצה קוד רק אם התנאי לא התקיים.</p>
</div>

<div style="text-align: right;" dir="rtl">
<h2>מבנה if / else</h2>
<p>דוגמה:</p>
</div>

```python
x = 5

if x > 0:
    print("Positive")
else:
    print("Negative")
```

<div style="text-align: right;" dir="rtl">
<p>x שווה ל-5, ולכן התנאי (x &gt; 0) הוא True, ולכן הודפס "Positive".</p>
<p>נמשיך לתרגל אופרטורים נוספים להשוואה:</p>
</div>

<div style="text-align: right;" dir="rtl">
<h2>אופרטור ==</h2>
<p>נחשו מה יודפס, ואז הריצו ובדקו:</p>
</div>

```python
x = 5

if x == 2:
    print("OK")
else:
    print("OOOOPS")
```

<div style="text-align: right;" dir="rtl">
<h2>אופרטורים &gt;= ו- &lt;=</h2>
<p>נחשו מה יודפס, ואז הריצו ובדקו:</p>
</div>

```python
x = 5

if x >= 2:
    print("OK")
else:
    print("OOOOPS")
```

<div style="text-align: right;" dir="rtl">
<h2>אופרטור != (שונה מ-)</h2>
<p>נחשו מה יודפס, ואז הריצו ובדקו:</p>
</div>

```python
x = 5

if x != 2:
    print("OK")
else:
    print("OOOOPS")
```

<h2>מספר תנאים - elif</h2>
<p>כאשר בוחנים כמה תנאים ברצף, אפשר להשתמש במילה <b>elif</b> (בין ה- if ל- else), ואפשר שיהיו כמה elif כאלו. ה- else מהווה את ברירת המחדל, כלומר מה שיקרה אם אף תנאי לא התקיים.</p>
<p>לדוגמה, נבדוק לפי כמות הזיכרון (memory) של טלפון לאיזו רמה הוא שייך. נחשו קודם מה יודפס עבור memory=4, ואז הריצו ובדקו:</p>
</div>

```python
memory = 4

if memory > 2:
    print("Medium level phone")
elif memory > 3:
    print("High end phone")
else:
    print("Low end phone")
```

<div style="text-align: right;" dir="rtl">
<p>הופתעתם מהתוצאה? למרות ש- memory=4 גדול גם מ-2 וגם מ-3, הודפס "Medium level phone".</p>
<p>הסיבה: פייתון בודקת את התנאים <b>לפי הסדר</b>, ומפסיקה ברגע שהיא מוצאת תנאי שהתקיים. מכיוון ש-4 &gt; 2 הוא כבר True, הקוד לא ממשיך לבדוק את ה- elif הבא.</p>
<p><b>מסקנה:</b> כשכותבים שרשרת של elif, חשוב לשים לב לסדר של התנאים!</p>
</div>

<h2>תנאים וקלט אקראי</h2>
<p>כדי להשתמש בקלט אקראי (מספר שנבחר במקרה), עלינו להוסיף בתחילת הקוד ספרייה שנקראת <b>random</b>.</p>
<p>לשם הדגמה, נשחק משחק ניחושים: המשתמש יבחר מספר בין 1 ל-7, והמחשב יבחר מספר אקראי באותו טווח. אם המספרים זהים - ניצחנו!</p>
</div>

```python
import random

num = int(input("choose a number between 1-7: "))
comp_num = random.randint(1, 7)

if num == comp_num:
    print("Bingo")
else:
    print("comp:", comp_num, " Try again")
```
